```python
from google.colab import files
import zipfile
import pandas as pd
import re
import os

# Upload archive.zip
uploaded = files.upload()

# Find the uploaded ZIP file
zip_path = next(name for name in uploaded if name.endswith(".zip"))

# Extract the required dataset
with zipfile.ZipFile(zip_path, "r") as z:
    csv_name = "aa_dataset-tickets-multi-lang-5-2-50-version.csv"
    z.extract(csv_name, "/content")

# Load dataset
df = pd.read_csv("/content/" + csv_name)

# Combine subject + body
df["combined_text"] = (
    df["subject"].fillna("") + " " +
    df["body"].fillna("")
)

# Clean text
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-ZÀ-ÿ\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["clean_text"] = df["combined_text"].apply(clean_text)

# Tokenization
df["tokens"] = df["clean_text"].apply(lambda x: x.split())

print("✅ Dataset restored!")
print("Shape:", df.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2)
)

X = tfidf.fit_transform(df["clean_text"])
y = df["queue"]

print("✅ TF-IDF features created!")
print("Feature matrix shape:", X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("✅ Train/Test split completed!")
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

print("✅ Model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("✅ Predictions completed!")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("🎯 Model Accuracy:", round(accuracy * 100, 2), "%")
print("\n📊 Classification Report:\n")
print(classification_report(y_test, y_pred))
```

```python
model_balanced = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

model_balanced.fit(X_train, y_train)

y_pred_balanced = model_balanced.predict(X_test)

print("✅ Balanced model trained and predictions completed!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy_balanced = accuracy_score(y_test, y_pred_balanced)

print("🎯 Balanced Model Accuracy:",
      round(accuracy_balanced * 100, 2), "%")

print("\n📊 Classification Report:\n")
print(classification_report(
    y_test,
    y_pred_balanced,
    zero_division=0
))
```

```python
def predict_ticket(subject, body):

    text = subject + " " + body

    # Clean text
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # Convert text into TF-IDF features
    text_vector = tfidf.transform([text])

    # Predict queue
    predicted_queue = model.predict(text_vector)[0]

    # Predict priority
    predicted_priority = priority_model.predict(text_vector)[0]

    print("🎫 SUPPORT TICKET PREDICTION")
    print("=" * 40)
    print("📝 Subject   :", subject)
    print("🏢 Queue     :", predicted_queue)
    print("🚦 Priority  :", predicted_priority.upper())
```

```python
# Error Analysis

errors = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

errors["Correct"] = errors["Actual"] == errors["Predicted"]

print("📊 ERROR ANALYSIS")
print("=" * 40)

print("\nTotal test tickets:", len(errors))
print("Correct predictions:", errors["Correct"].sum())
print("Incorrect predictions:", (~errors["Correct"]).sum())

print("\n❌ Most common prediction errors:")
print(
    errors[~errors["Correct"]]
    .groupby(["Actual", "Predicted"])
    .size()
    .sort_values(ascending=False)
    .head(10)
)
```
